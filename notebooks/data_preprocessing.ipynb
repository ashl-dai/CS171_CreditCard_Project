## **Data Exploration and Data Preprocessing**


### **Data Cleaning**

```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
```

```python
# importing the data
credit_card_data = fetch_ucirepo(id=350)
X = credit_card_data.data.features # input features
y = credit_card_data.data.targets  # target
```

```python
# inital check to see if data is read in correctly
X.head()
```

```python
y.head()
```

```python
# check dimensions
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# checking type and missing values
credit_card_data.variables
```

```python
# rename column names to more descriptive names
X.columns = [
    "LIMIT_BAL", "SEX", "EDUCATION", "MARRIAGE", "AGE",
    "PAY_1", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6",
    "BILL_AMT1", "BILL_AMT2", "BILL_AMT3", "BILL_AMT4",
    "BILL_AMT5", "BILL_AMT6",
    "PAY_AMT1", "PAY_AMT2", "PAY_AMT3", "PAY_AMT4",
    "PAY_AMT5", "PAY_AMT6"
]
y.columns = ['default_payment_next_month']

# verifying changes
print(X.columns.tolist())
print(y.columns.tolist())
```

```python
# checking the values of our categorical variables
categorical_cols = [
    "SEX",
    "EDUCATION",
    "MARRIAGE",
    "PAY_1",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]
for col in categorical_cols:
    print(f"{col}: {sorted(X[col].unique().tolist())}\n")
```

We will follow what each value means based on this post: https://www.kaggle.com/datasets/uciml/default-of-credit-card-clients-dataset/discussion/34608 

```python
# MARRIAGE (Marital Status): 1=married, 2=single, 3=divorce, 0=others
# PAY_1-6: -2: No consumption; -1: Paid in full; 0: The use of revolving credit; 1 = payment delay for one month; 2 = payment delay for two months; . . .; 8 = payment delay for eight months; 9 = payment delay for nine months and above.

# EDUCATION: values 0,4,5,6 means "other". Mapping all these values to 4 for consistency
X["EDUCATION"] = X["EDUCATION"].replace({
    0: 4,
    5: 4,
    6: 4
})
# verify
print(f"EDUCATION: {X["EDUCATION"].unique()} \n")
```

```python
# change categorical variables typing to categorical
for col in categorical_cols:
    X[col] = X[col].astype('category')
    
# verifying changes
X.info()
```

```python
# checking for unusual values
X.describe()
```

```python
# BILL_AMT1-6 has negative minimums
bill_cols = [
    "BILL_AMT1", "BILL_AMT2", "BILL_AMT3", 
    "BILL_AMT4", "BILL_AMT5", "BILL_AMT6"
]
for col in bill_cols:
    print(f"{col}: {(X[col] < 0).sum()} negative values")
```

```python
for col in bill_cols:
    print(f"\n{col}")
    print(X.loc[X[col] < 0, col].describe())
```

**BILL_AMT1-6**: Keeping the negative values because it could represent refunds. Looking at the distributions, specifically at the median we can see that most of the negative values are not super far off from 0. There is some negative billing amount outliers but they are not extreme enough to discard as invalid data. Reminder that these billing amounts are in Taiwan's currency.

```python
# check for duplicates
credit_card_df = pd.concat([X, y], axis=1)
print(credit_card_df.duplicated().sum())
```

```python
credit_card_df = credit_card_df.drop_duplicates()

X = credit_card_df.drop(columns=['default_payment_next_month'])
y = credit_card_df[['default_payment_next_month']]

# verifying the duplicates are dropped
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```
