## **Data Exploration**


```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# importing the data
credit_card_data = fetch_ucirepo(id=350)
X = credit_card_data.data.features # input features
y = credit_card_data.data.targets  # target
```

```python
# inital check to see if data is read in correctly
X.head()
```

```python
y.head()
```

```python
# check dimensions
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# checking type and missing values
credit_card_data.variables
```

```python
# rename column names to more descriptive names
X.columns = [
    "LIMIT_BAL", "SEX", "EDUCATION", "MARRIAGE", "AGE",
    "PAY_1", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6",
    "BILL_AMT1", "BILL_AMT2", "BILL_AMT3", "BILL_AMT4",
    "BILL_AMT5", "BILL_AMT6",
    "PAY_AMT1", "PAY_AMT2", "PAY_AMT3", "PAY_AMT4",
    "PAY_AMT5", "PAY_AMT6"
]
y.columns = ['default_payment_next_month']

# verifying changes
print(X.columns.tolist())
print(y.columns.tolist())
```

```python
# checking the values of our categorical variables
categorical_cols = [
    "SEX",
    "EDUCATION",
    "MARRIAGE",
    "PAY_1",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]
for col in categorical_cols:
    print(f"{col}: {sorted(X[col].unique().tolist())}\n")
```

We will follow what each value means based on this post: https://www.kaggle.com/datasets/uciml/default-of-credit-card-clients-dataset/discussion/34608 

```python
# MARRIAGE (Marital Status): 1=married, 2=single, 3=divorce, 0=others
# PAY_1-6: -2: No consumption; -1: Paid in full; 0: The use of revolving credit; 1 = payment delay for one month; 2 = payment delay for two months; . . .; 8 = payment delay for eight months; 9 = payment delay for nine months and above.

# EDUCATION: values 0,4,5,6 means "other". Mapping all these values to 4 for consistency
X["EDUCATION"] = X["EDUCATION"].replace({
    0: 4,
    5: 4,
    6: 4
})
# verify
print(f"EDUCATION: {X["EDUCATION"].unique()} \n")
```

```python
# change categorical variables typing to categorical
for col in categorical_cols:
    X[col] = X[col].astype('category')
    
# verifying changes
X.info()
```

```python
# checking for unusual values
X.describe()
```

```python
# BILL_AMT1-6 has negative minimums
bill_cols = [
    "BILL_AMT1", "BILL_AMT2", "BILL_AMT3", 
    "BILL_AMT4", "BILL_AMT5", "BILL_AMT6"
]
for col in bill_cols:
    print(f"{col}: {(X[col] < 0).sum()} negative values")
```

```python
for col in bill_cols:
    print(f"\n{col}")
    print(X.loc[X[col] < 0, col].describe())
```

```python
numerical_cols = X.select_dtypes(include='number').columns

for col in numerical_cols:
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = X[
        (X[col] < lower_bound) |
        (X[col] > upper_bound)
    ]

    print(f'{col}: {len(outliers)} outliers')
```

**BILL_AMT1-6**: Keeping the negative values because it could represent refunds. Looking at the distributions, specifically at the median we can see that most of the negative values are not super far off from 0. There is some negative billing amount outliers but they are not extreme enough to discard as invalid data. Reminder that these billing amounts are in Taiwan's currency. 

**Overall Outliers:** Furthermore, the outliers for bill amount and pay amount could be useful predictive information because large balances or payments could be correlated with default behavior.

```python
# check for duplicates
credit_card_df = pd.concat([X, y], axis=1)
print(credit_card_df.duplicated().sum())
```

```python
credit_card_df = credit_card_df.drop_duplicates()

X = credit_card_df.drop(columns=['default_payment_next_month'])
y = credit_card_df[['default_payment_next_month']]

# verifying the duplicates are dropped
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Univariate Analysis
plt.figure(figsize=(10, 5))

sns.histplot(data=X, x='BILL_AMT1', bins=50)

plt.title('Distribution of First Month Billing Amount')
plt.xlabel('Billing Amount')
plt.ylabel('Number of Customers')
plt.show()
```

**Distribution:** The data is heavily skewed to the right but because we are using XGBoost model, we decided to not transform the data. This is because XGBoost uses decision trees which makes splits based on feature values and their ordering instead of the overall distribution.

```python
# Bivariate Analysis
target_col = "default_payment_next_month"
plot_nums = ["LIMIT_BAL", "AGE", "BILL_AMT1", "PAY_AMT1"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for idx, col in enumerate(plot_nums):
    sns.boxplot(
        data=credit_card_df,
        x=target_col,
        y=col,
        showfliers=False,
        ax=axes[idx]
    )
    axes[idx].set_title(f"{col} by {target_col}")
    axes[idx].set_xlabel(target_col)

plt.tight_layout()
plt.show()
```

**Bivariate Analysis:** For limit balance, defaulters (1) have lower credit limits than the non-defaulters (0). For age, the distribution is roughly the same for defaulters and non-defaulters. For bill amount 1, they are also around the same. For pay amount 1, defaulters make smaller recent payments than non-defaulters.

```python
# Categorical frequencies check
n_cols = 3
n_rows = (len(categorical_cols) + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, n_rows * 3.5))
axes = axes.flatten()

for idx, col in enumerate(categorical_cols):
    sns.countplot(
        data=credit_card_df,
        x=col,
        order=sorted(credit_card_df[col].dropna().unique()),
        ax=axes[idx]
    )
    axes[idx].set_title(f"{col} Frequency")
    axes[idx].set_xlabel("")
    axes[idx].tick_params(axis="x")

for ax in axes[len(categorical_cols):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

**Categorical Frequencies**: For sex, there is a noticeably less males (1) than females (2). For education, most individuals have a university (2) or graduate (1) level while high school and "other" are the minority. For marital status, most people tend to be married (1) or single (2), in this dataset very little amount of people represent divorced (3). For the pay_1-6, there is a dispportionate amount of people in this dataset that used revolving credit, most people dont seem to to delay payment more than 3 months.

```python
target_col = "default_payment_next_month"

fig, ax = plt.subplots(figsize=(6, 4))

sns.countplot(
    data=credit_card_df,
    x=target_col,
    order=sorted(credit_card_df[target_col].dropna().unique()),
    ax=ax
)

ax.set_title(f"{target_col} Frequency")
ax.set_xlabel("")
ax.set_xticks([0, 1])
ax.set_xticklabels(["Non-Defaulters (0)", "Defaulters (1)"])
ax.tick_params(axis="x")

plt.tight_layout()
plt.show()
```

**Categorical Frequencies**: There is a disproportionately higher number of non-defaulters (0) than defaulters (1) in this dataset.
